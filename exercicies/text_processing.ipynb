```python
# import libs
import nltk

# import data 
iliad_path = "../data/the-Iliad-of-homer.txt"
with open(iliad_path, "r") as f:
    iliad = f.read()
```

# 1. Tokenização (word)
• Saída esperada: Uma lista que contém strings representando os tokens extraídos.

```python
from nltk.tokenize import word_tokenize

def print_tokens_count(tokens):
    print("--------------------------------")
    print(f'Total of tokens: %s' % len(tokens))

def print_all_tokens(tokens):
    print_tokens_count(tokens)
    print("--------------------------------\nAll tokens:")
    
    text = ""
    for token in tokens:
        text += token
        text += " "

    print(text)

def list_all_tokens(tokens):
    print_tokens_count(tokens)
    print("--------------------------------\nAll tokens:")
    
    for token in tokens:
        print(token)

tokens = word_tokenize(iliad)
print_all_tokens(tokens)
```

# 2. Normalização e Limpeza.

## (a) Lower casing (conversão de todos os caracteres para minúsculo);

```python
lower_tokens = []
for token in tokens:
    lower_tokens.append(token.lower())
    
print_all_tokens(lower_tokens)
```

## (b) Remoção de pontuações e outros caracteres especiais;

```python

def only_worlds(tokens):
    words = []
    for token in tokens:
        if(token.isalpha()):
            words.append(token)

    return words

words = only_worlds(tokens)
    
print_all_tokens(words)
```

## (c) Remoção de stopwords;

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

def remove_stopwords(tokens):
    filtered_words = []
    for token in tokens:
        if(not token in stop_words):
            filtered_words.append(token)

    return filtered_words

filtered_words = remove_stopwords(words)
    
print_all_tokens(filtered_words)
```

## (d) Remoção das palavras mais frequentes;

### d.1 - Implemente a remoção das n palavras mais frequentes

```python
def print_words_frequency(frequent_words):
    print("--------------------------------")
    print("Frequency of words:")
    for word in frequent_words:
        print(word[0] + " - " + str(word[1]))

def remove_frequent_words(tokens, num_words=3):
    # filtering words
    words = only_worlds(tokens)
    filtered_words = remove_stopwords(words)

    # getting the frequency
    freq = nltk.FreqDist(filtered_words)
    frequent_words_dict = freq.most_common(num_words)
    print_words_frequency(frequent_words_dict)

    frequent_words = [word[0] for word in frequent_words_dict]
    total = len(filtered_words)
    removed = 0
    # removing the most frequent words
    for token in tokens:
        if(token in frequent_words):
            removed+=1
            filtered_words.remove(token)

    print("--------------------------------")
    print(f"Removed {removed} words of {total}.")
    return filtered_words

not_frequent_words = remove_frequent_words(tokens)
print_all_tokens(not_frequent_words)
```

### d.2 - Remoção das palavras que ocorreram mais que n vezes, sendo n um argumento para a função.

```python
def remove_words_by_most_occurrence(tokens, num_occ=3):
    # filtering words
    words = only_worlds(tokens)
    filtered_words = remove_stopwords(words)

    # getting the frequency
    freq = nltk.FreqDist(filtered_words)
    frequent_words_dict = [item for item in freq.items() if item[1] >= num_occ]
    # print_words_frequency(frequent_words_dict)

    frequent_words = [word[0] for word in frequent_words_dict]

    total = len(filtered_words)
    removed = []
    # removing the most frequent words
    for token in tokens:
        if(token in frequent_words):
            removed.append(token)
            filtered_words.remove(token)

    print("--------------------------------")
    print(f"Removed {len(removed)} words of {total}.")
    return filtered_words

frequent_words_by_occurrence_words = remove_words_by_most_occurrence(tokens, 400)
print_all_tokens(frequent_words_by_occurrence_words)
```

# 4.  Faça um ranking dos 20 types mais frequentes no texto, mostrando também a frequência de cada um deles;

```python
def complete_filter(tokens):
    words = only_worlds(tokens)
    filtered_words = remove_stopwords(words)

    return filtered_words
    

def print_words_frequency(frequent_words):
    print("--------------------------------")
    print("Frequency of words:")
    for word in frequent_words:
        print(word[0] + " - " + str(word[1]))

def list_most_common_types(tokens, num_tokens = 20):
    # getting the frequency
    freq = nltk.FreqDist(tokens)
    frequent_words_dict = freq.most_common(num_tokens)
    print_words_frequency(frequent_words_dict)

print("--------------------------------")
print(f"Non filtered tokens: {len(tokens)}")
list_most_common_types(tokens)

filtered = complete_filter(tokens)
print("--------------------------------")
print(f"Filtered tokens: {len(filtered)}")
list_most_common_types(filtered)
```
