# Desenvolva um pequeno e simples corretor ortográfico


## Entrada: texto
## Saida: lista de palavras ortograficamente erradas, com sugestão de correção para cada uma delas.
## Vocabulário:
> ["O" "A" "Os" "As" "homem" "homens" "mulher" "mulheres" "é" "são" "inteligente" "inteligentes" "brilhante" "brilhantes"]

## Frases de Exemplo:
  - o homem é inteligentr
  - Os homens sõ brilhantess
  - As muleres são brilantes
  - A milher é inteliggente

# 1º passo
Calcular a distância de edição mínima de Leveshtein

```python
def levenshtein_distance(word1, word2):
    rows = len(word1) + 1
    cols = len(word2) + 1
    distance_matrix = [[0 for _ in range(cols)] for _ in range(rows)]

    for i in range(1, rows):
        distance_matrix[i][0] = i
    for j in range(1, cols):
        distance_matrix[0][j] = j

    for i in range(1, rows):
        for j in range(1, cols):
            if word1[i - 1] == word2[j - 1]:
                cost = 0
            else:
                cost = 1
            distance_matrix[i][j] = min(
                distance_matrix[i - 1][j] + 1,
                distance_matrix[i][j - 1] + 1,
                distance_matrix[i - 1][j - 1] + cost
            )

    return distance_matrix[-1][-1]
```

## 2º passo
Reconhecer as palavras que não estão no vocabulário e substituir pela palavra que mais se parece com ela.

```python
def spell_checker(sentence, vocabulary):
    words = sentence.split()
    vocabulary = [word for word in vocabulary]
    
    misspelled_words = []
    
    for word in words:
        if word not in vocabulary:
            closest_match = min(vocabulary, key=lambda v: levenshtein_distance(word, v))
            if levenshtein_distance(word, closest_match) < 3: 
                misspelled_words.append((word, closest_match))
    
    return misspelled_words
```

## 3º passo
Testar a solução

```python
vocabulary = [
    "O", "A", "Os", "As", "homem", "homens", "mulher", "mulheres",
    "é", "são", "inteligente", "inteligentes", "brilhante", "brilhantes"
]

sentences = [
    "o homem é inteligentr",
    "Os homens sõ brilhantess",
    "As muleres são brilantes",
    "a milher é inteliggente"
]

suggestions = []

for sentence in sentences:
    misspelled_words = spell_checker(sentence, vocabulary)

    sentence_suggestion = f"Correções para a frase '{sentence}':\n"
    
    if misspelled_words:
        for word, correction in misspelled_words:
            sentence_suggestion += f"'{word}' -> '{correction}'\n"
    else:
        sentence_suggestion += "Nenhum erro encontrado.\n"

    suggestions.append(sentence_suggestion)

for suggestion in suggestions:
    print(suggestion)
```
