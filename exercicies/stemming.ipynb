Considerando as bases de texto fornecidas na atividade 2, implementem as seguintes atividades, que
correspondem a algumas tarefas comuns em um pipeline de pré-processamento de texto:

````python
# import libs
import nltk
from nltk import pos_tag
from nltk import word_tokenize
from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer, LancasterStemmer, PorterStemmer
from utils import print_words_frequency, print_all_tokens, remove_stopwords, only_words, print_bar
````

````python
# import data 
iliad_path = "../data/the-Iliad-of-homer.txt"
with open(iliad_path, "r") as f:
    iliad = f.read()

iliad_tokens = word_tokenize(iliad)
````

# 1. Lematização.
    • Aplique um método de lematização em uma lista de tokens;
    • Pode-se considerar uma lista tokens da língua inglesa;
    • Saída esperada: Uma lista dos tokens após o processo de lematização.


##  Explicação:

> Para reduzir uma palavra ao seu lema, foi se verificado a <i>tag</i> da palavra, isto é, sua classe gramatical. Após ter conhecimento da classe gramatical, basta importar o <i>corpus</i> referente ao método utilizado (<i>wordnet</i>) e fazer a conversão para cada palavra passando como parâmetro sua classe. Fica perceptivel que palavras como "<i>done</i>" e "<i>are</i>" se tornam "<i>do</i>" e "<i>be</i>".

````python
def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

def apply_lemmatization(tokens):
    lemmatizer = WordNetLemmatizer()
    pos_tags = pos_tag(tokens)

    lemmatized_words = [lemmatizer.lemmatize(word, get_wordnet_pos(pos)) for word, pos in pos_tags]

    return lemmatized_words

iliad_lematized = apply_lemmatization(iliad_tokens)

print_all_tokens(iliad_tokens)
print_all_tokens(iliad_lematized)
````

# 2. Stemming.
    • Aplique um método de stemming em uma lista de tokens;
    • Pode-se considerar uma lista tokens da língua inglesa ou da lingua portuguesa;
    • Saída esperada: Uma lista dos tokens após o processo de stemming.

 - Usar pelo menos dois métodos e comparar os resultados



##  Explicação:
> O processo de <i>stemming</i> pode ser aplicado por diversos algoritmos, para este estudo foram selecionados dois métodos super populares: [Porter](https://doi.org/10.1108/eb046814) e [Lancaster](https://dl.acm.org/doi/pdf/10.5555/188490.188499). Com os métodos aplicados, torna-se possível realizar uma comparação e assim perceber que o método de Lancaster é muito mais radical de que o de Porter, pois corta quase todo tipo de palavra, se baseando apenas em sufixos e prefixos. Já o método de Porter aparenta realizar a redução para de o radical de forma mais consciente, focando em morfemas conhecidos e clássicos da linguagem, como a remoção de "<i>ing</i>".

````python
def apply_porter_stemming(tokens):
    stemmer = PorterStemmer()
    return [stemmer.stem(token) for token in tokens]

def apply_lancaster_stemming(tokens):
    stemmer = LancasterStemmer()
    return [stemmer.stem(token) for token in tokens]
    
iliad_after_porter = apply_porter_stemming(iliad_tokens)
iliad_after_lancaster = apply_lancaster_stemming(iliad_tokens)

print_bar()
print("Texto original")
print_all_tokens(iliad_tokens)

print_bar()
print()

print("Após o Porter")
print_all_tokens(iliad_after_porter)

print_bar()
print()

print("Após o Lancaster")
print_all_tokens(iliad_after_lancaster)
print_bar()
````

# 3. Faça um ranking dos 20 lemas e das 20 raízes mais frequentes no texto, novamente mostrando a frequência de cada um deles.

````python
no_symbols = only_words(iliad_tokens)
filtered_tokens = remove_stopwords(no_symbols)

iliad_lematized = apply_lemmatization(filtered_tokens)
iliad_after_lancaster = apply_lancaster_stemming(filtered_tokens)

freq = nltk.FreqDist(filtered_tokens)
frequent_dict = freq.most_common(20)

lemmas_freq = nltk.FreqDist(iliad_lematized)
frequent_lemmas_dict = lemmas_freq.most_common(20)

stem_freq = nltk.FreqDist(iliad_after_lancaster)
frequent_stem_dict = stem_freq.most_common(20)

print("Lemmas")
print_words_frequency(frequent_lemmas_dict)

print_bar()
print("Raizes")
print_words_frequency(frequent_stem_dict)

print_bar()
print("Types")
print_words_frequency(frequent_dict)
````

## Distância de Levenshtein
Consiste em criar uma matrix com o valor de todas trocas entre duas palavras e retornar o valor final sendo equivalente a última posição desta matrix, como pode ser visto no pseudocódigo abaixo:

```pseudo code
função distancia_minima_de_edição(palavra1, palavra2):
    altura <- tamanho(palavra1)
    comprimento <- tamanho(palavra2)
    
    matriz_comparação <- matriz(altura, comprimento)
    
    repita i de 0 até altura:
        dp[i][0] <- i 
        
    repita j de 0 até comprimento:
        dp[0][j] <- j

    repita i de 1 até altura:
        para j de 1 até comprimento:
            se palavra1[i-1] igual palavra2[j-1]:
                custo <- 0 
            caso não:
                custo <- 1  

            deletar <- dp[i-1][j] + 1
            inserir <- dp[i][j-1] + 1
            substituir <- dp[i-1][j-1] + custo

            dp[i][j] = valor_minimo(
                deletar,       
                inserir,       
                substituir 
            )
    
    retorno matriz_comparação[altura][comprimento]
```
