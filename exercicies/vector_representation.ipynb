# Vector Representation of Words/Documents
Considerando as bases de texto fornecidas, implemente as seguintes atividades, que correspondem a algumas tarefas comuns em um pipeline de pré-processamento de texto:

### Pre process text

```python
import numpy as np

# import data 
iliad_path = "../data/the-Iliad-of-homer.txt"
with open(iliad_path, "r") as f:
    iliad = f.read()


from nltk import word_tokenize

def preprocess_text(corpus):
    tokens = word_tokenize(corpus)
    words = []
    for token in tokens:
        if(token.isalpha()):
            words.append(token)

    return  " ".join(words)

iliad_corpus = preprocess_text(iliad)
```

## 1. Representação vetorial de textos.
Sua função deve extrair o vocabulário do Corpus e gerar vetores (com o tamanho do vocabulário) que representam cada texto considerando as seguintes abordagens:

### (a) One Hot Encode - Binária, onde cada posição dos vetores deve indicar se uma palavra ocorreu ou não em um texto;

```python
import numpy as np

def one_hot_encode(corpus):
  token_index = {}
  counter = 0
  
  for considered_word in corpus.split():
    if considered_word not in token_index:
      token_index.update({considered_word : counter + 1}) 
      
      counter = counter + 1   

  max_length = max([len(sample.split()) for sample in corpus])
  
  results  = np.zeros(shape = (len(corpus),
                              max_length,
                              max(token_index.values()) + 1)) 
  
  for i, sample in enumerate(corpus): 
    for j, considered_word in list(enumerate(sample.split())):
      
      index = token_index.get(considered_word)
      
      results[i, j, index] = 1. 

  return results, token_index

def print_encoding_result(result, token_dict):
  print(f'Palavras únicas: {token_dict}')
  print(f'Tamanho do tensor: {result.shape}')
  print(f'Corpus após o enconding: {result}')

iliad_encoded, iliad_dict = one_hot_encode(iliad_corpus)
print_encoding_result(iliad_encoded, iliad_dict)
```

```python
# Exemplo mais simples
second_corpus = "Xuxa, a sasha fez xixi na cama"

second_encoded, second_dict = one_hot_encode(second_corpus)
print_encoding_result(second_encoded, second_dict)
```

### (b) One Hot Encode - Contagem, onde cada posição dos vetores deve indicar o número de vezes que uma palavra ocorreu no texto;

```python
def frequency_encode(corpus):
    token_index = {}
    counter = 0

    # Build token index (vocabulary)
    for sample in corpus:                                      
        for considered_word in sample.split():
            if considered_word not in token_index:
                token_index[considered_word] = counter + 1  # Assign unique index to each word
                counter += 1

    # Create a tensor to hold word frequency encoding (num_samples, num_unique_words)
    results = np.zeros(shape=(len(corpus),
                              max(token_index.values()) + 1))

    # Perform frequency encoding
    for i, sample in enumerate(corpus):
        for considered_word in sample.split():
            index = token_index.get(considered_word)
            # Increment the frequency count for the word in the current sample
            results[i, index] += 1

    return results, token_index

# Apply frequency encoding to the Iliad corpus
iliad_frequency_encoded, iliad_frequency_dict = frequency_encode(iliad_corpus)
print_encoding_result(iliad_frequency_encoded, iliad_frequency_dict)
```

```python

second_corpus = ["Xuxa, a sasha fez xixi na cama"]

second_frequency_encoded, second_frequency_dict = frequency_encode(second_corpus)
print_encoding_result(second_frequency_encoded, second_frequency_dict)
```
