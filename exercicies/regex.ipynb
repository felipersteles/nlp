# 1.Escreva expressões regulares para os seguintes padrões

## a) Todas as strings alfabéticas;

```python
from nltk.tokenize import RegexpTokenizer

text = "Jamais o sol_.! veê a sombra!?. - Leonardo da 20 Vinci"

tokenizer = RegexpTokenizer(r'[A-Za-z]+')
res = tokenizer.tokenize(text)

print(text)
print(res)
```

## b) Todas as strings alfabéticas minúsculas terminando em a b;

```python
from nltk.tokenize import RegexpTokenizer

text = "O jabuticabeiral exuberante abrigavab sabiás que sabiamente elaboravam melodias, enquanto o caboclo, com habilidade, elaborab uma rede de zabumba, embalando a cabanab com um som suave."

tokenizer = RegexpTokenizer(r'\b[a-z]+ab\b')
res=tokenizer.tokenize(text)

print(text)
print(res)
```

## c) Todas as strings do alfabeto a; b de modo que cada a seja imediatamente precedido por e imediatamente seguido por a b;

```python
from nltk.tokenize import RegexpTokenizer

text = "Os marabaenses celebram suas raízes com muita alegria, preservando as tradições e a língua tupi."

tokenizer = RegexpTokenizer(r'\b[A-Za-z]*abae[A-Za-z]*\b')
res=tokenizer.tokenize(text)

print(text)
print(res)
```

## d) Corresponder à palavra "viagra" e a alguns dos disfarces que remetentes de spam usam, como: vi@gra ; v1agra; v1@gra; v!@gr@

```python
from nltk.tokenize import RegexpTokenizer

text = "vi@gra ; v1agra; v1@gra; v!@gr@ e outras palavras para realização de testes em experimentos"

tokenizer = RegexpTokenizer(r'\bv[i1!][a@]gr[a@]+')
res=tokenizer.tokenize(text)

print(text)
print(res)
```

## e) Corresponder a qualquer endereço de e-mail dos domínios nca.ufma e ufpi.edu.br

```python
from nltk.tokenize import RegexpTokenizer

text = "paiva@nca.ufma.br, faps@ufpi.edu.br, example@email.com, testes@testes.com, teste@ufpi.edu.br teste@nca.ufma.br"

tokenizer = RegexpTokenizer(r'\b[A-Za-z0-9]+@(?:nca\.ufma\.br|ufpi\.edu\.br)\b')
res = tokenizer.tokenize(text)

print(text)
print(res)
```

## f) Corresponder a qualquer endereço IP com a variação de 192.168.1.0 a 192.168.1.255.

```python
from nltk.tokenize import RegexpTokenizer

text = '192.168.1.0 - 192.168.3.100 - 192.168.1.255 -  192.168.1.300 - 192.168.1.200 - 192.168.1.3 - 192.168.5.13 - 219.186.1.5 - 921.681.1.90'

tokenizer = RegexpTokenizer(r'192\.168\.1\.[0-255]{1,3}')
res = tokenizer.tokenize(text)

print(text)
print(res)
```

## g) Corresponder aos números de ordem de compra (PO) de sua empresa. Esse número tem vários formatos possíveis, como: PO nn-nnnnn; PO-nn-nnnn; PO#nn nnnn; PO#nn-nnnn;  PO nnnnnn

```python
from nltk.tokenize import RegexpTokenizer

text = 'PO 97-132512 - PO-43-1982 - PO#12 1281 - PO#11-442222244 - PO#11-4444 - PO#11-44421sss14 - PO 192819 - testestestes'

tokenizer = RegexpTokenizer(r'PO[\s#-][0-9]{2}[\s#-]?[0-9]{4,6}')
res=tokenizer.tokenize(text)

print(text)
print(res)
```

# 2. Escreva expressões regulares para os seguintes padões. Por “palavra”, queremos dizer uma string alfabética separada de outras palavras por espaços em branco, qualquer pontuação relevante, quebras de linha e assim por diante.

## a)Todas as strings com duas palavras repetidas consecutivas (por exemplo, “Humbert Humbert” e “o o”);

```python
from nltk.tokenize import RegexpTokenizer

text = 'Humbert Humbert o o testestes testestes testest esteste telescopio magico'

tokenizer = RegexpTokenizer(r'\b(\w+)\s+\1\b')
res=tokenizer.tokenize(text)

print(text)
print(res)
```


## b)Todas as strings que começam no início da linha com um inteiro e terminam no final da linha com uma palavra;


```python
from nltk.tokenize import RegexpTokenizer

text = """
1. Pré-aqueça o forno a 325 graus.
2. Unte uma assadeira 8 x 8
3. Misture todos os ingredientes secos (listados acima).
4. Delicadamente, junte os ingredientes úmidos (listados acima).
5. Despeje a massa na assadeira e leve ao forno por 45 minutos
"""

lines = text.split('\n')

tokenizer = RegexpTokenizer(r'^\d+.*\w+$')
matching_lines = [line for line in lines if tokenizer.tokenize(line)]

print(text)
print(matching_lines)
```

## c) Todas as strings que contêm a palavra gruta e a palavra raven (mas não, por exemplo, palavras como grutas que contêm apenas a palavra gruta);

```python
from nltk.tokenize import RegexpTokenizer

text = "No fundo da gruta grutas raven só havia água e pedra. E peixes albinos, que acreditavam que gruta o mundo era o lago onde viviam. O Peixinho raven ouvia as histórias que o avô contava e não acreditava que gruta pudessem ser verdade. Não havia vida fora dali. raven"

tokenizer = RegexpTokenizer(r'\b(raven|gruta)\b')
res = tokenizer.tokenize(text)

print(text)
print(res)
```
